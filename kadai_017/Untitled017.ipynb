```python
# python 第１７章　課題(kadai_017) テクニックを組み合わせてコードを記述
#
# クラス　名前と年齢を受け取り、大人か否かの判定
class Human:
  # コンストラクタで名前、年齢の情報を設定
  def __init__(self,name,age):
    self.name = name
    self.age = age

  # 年齢で大人か否かの判断を行い、表示するメソッド
  def check_adult(self):

    if self.age < 20:
      check = "大人では、ありません。"
    else:
      check = "大人です。"

    print(f"{self.name}さんは{self.age}歳。{check}")

# 複数人のデータを設定
person ={"安藤さくら":19,
         "伊藤太郎":20,
         "鵜飼さぶろう":35,
         "江崎ももこ":29,
         "岡本史郎":18,
         "山田太郎":9,
         "加賀美ゆい":21}

# 設定した人数分、繰り返し
for person_name,person_age in person.items():

  # クラスに情報を渡し、メソッドの実行
  human = Human(person_name,person_age)
  human.check_adult()

```
