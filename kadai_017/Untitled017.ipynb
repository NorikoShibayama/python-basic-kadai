```python
# python 第１７章　課題(kadai_017) テクニックを組み合わせてコードを記述
#
# クラス　名前と年齢を受け取り、大人か否かの判定
class Human:
  # コンストラクタで名前、年齢の情報を設定
  def __init__(self,name,age):
    self.name = name
    self.age = age

  # 年齢で大人か否かの判断を行い、表示するメソッド
  def check_adult(self):

    if self.age < 20:
      check = "大人では、ありません。"
    else:
      check = "大人です。"

    print(f"氏名：{self.name}　年齢：{self.age}・・{check}")

# インスタンス作成し humansリスト作成
humans = []
humans.append(Human("安藤さくら",19))
humans.append(Human("伊藤太郎",20))
humans.append(Human("鵜飼さぶろう",30))
humans.append(Human("江崎四郎",21))
humans.append(Human("小川ももこ",18))
humans.append(Human("安藤さくら",9))

# humansリストから出力する
for person in humans:

  human = Human(person.name,person.age)
  human.check_adult()

```
